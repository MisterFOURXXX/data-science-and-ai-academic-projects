# Approach

**Preprocess Pipeline Approach**

The preprocessing pipeline transforms raw text into a clean, numerical format suitable for the Transformer model. The steps include:
- **Data Cleaning**: Remove missing values and duplicates to ensure data quality.
- **Text Preprocessing**:
  - Convert text to lowercase for consistency.
  - Remove punctuation and numbers using regex, retaining only alphabetic characters.
  - Tokenize text into words using whitespace splitting.
  - Remove stopwords (e.g., "the", "is") using NLTK's stopword list.
  - Apply Porter stemming to reduce words to their root form (e.g., "running" to "run").
- **TF-IDF Vectorization**: Convert text to TF-IDF vectors, limiting to 5,000 features to capture word importance while managing dimensionality.
- **Dimensionality Reduction**: Apply Truncated SVD to reduce TF-IDF vectors to 128 dimensions for computational efficiency.
- **Label Encoding**: Use `LabelEncoder` to convert categorical labels into integers (0 to 3).

**Theories and Concepts**

- **TF-IDF (Term Frequency-Inverse Document Frequency)**: A statistical measure that quantifies the importance of a word in a document relative to the entire corpus. Term Frequency (TF) measures how often a word appears in a document, while Inverse Document Frequency (IDF) reduces the weight of common words across documents. The TF-IDF score is calculated as:
  $$
  \text{TF-IDF}(t, d, D) = \text{TF}(t, d) \cdot \text{IDF}(t, D), \quad \text{IDF}(t, D) = \log\left(\frac{|D|}{\text{DF}(t, D)}\right)
  $$
  where $t$ is the term, $d$ is the document, $D$ is the corpus, and $\text{DF}(t, D)$ is the number of documents containing $t$. This approach highlights discriminative words for classification.
- **Tokenization**: Splitting text into individual words or tokens to enable further processing.
- **Stopwords**: Common words with low semantic value, removed to reduce noise in the feature set.
- **Stemming**: Reducing words to their root form to normalize variations (e.g., "runs", "running" to "run"), improving feature consistency.
- **Truncated SVD**: A dimensionality reduction technique that approximates the TF-IDF matrix using singular value decomposition, reducing computational complexity while preserving key variance:
  $$
  A \approx U_k \Sigma_k V_k^T
  $$
  where $A$ is the TF-IDF matrix, and $k$ is the number of components (128 in this case).

**Model Training Approach**

The model combines TF-IDF features with a Transformer architecture implemented in TensorFlow/Keras. Two implementations are provided: one using Keras's built-in layers and another with custom layers for flexibility.

**Keras Transformer**

- **Architecture**:
  - **Input Layer**: Accepts TF-IDF vectors (shape: [batch_size, 128]).
  - **Reshape**: Expands dimensions to treat the TF-IDF vector as a sequence of length 1 (shape: [batch_size, 1, 128]).
  - **Multi-Head Attention**: Uses `MultiHeadAttention` with 4 heads to capture relationships between features.
  - **Feed-Forward Network**: Applies a dense layer with ReLU activation (128 units) for non-linearity.
  - **Residual Connections and Normalization**: Adds input to attention and feed-forward outputs, followed by `LayerNormalization` to stabilize training.
  - **Global Pooling**: Uses `GlobalAveragePooling1D` to reduce the sequence dimension.
  - **Classification Head**: Includes dense layers (64 units with ReLU, dropout 0.2, and 4 units with softmax) for final predictions.
- **Compilation**: Uses Adam optimizer (learning rate 0.001) and `SparseCategoricalCrossentropy` loss for integer labels.
- **Training**: Trains for up to 200 epochs with early stopping (patience=5) based on validation loss to prevent overfitting.

**Custom Transformer**

- **MultiHeadSelfAttention Layer**:
  - Projects inputs into query, key, and value vectors using dense layers.
  - Splits projections into multiple heads for parallel attention.
  - Computes scaled dot-product attention:
    $$
    \text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
    $$
    where $Q$, $K$, and $V$ are query, key, and value matrices, and $d_k$ is the key dimension.
  - Concatenates heads and applies a final dense layer.
- **TransformerBlock Layer**:
  - Combines `MultiHeadSelfAttention`, a feed-forward network (dense layers with ReLU), dropout, and layer normalization.
  - Uses residual connections to add inputs to outputs, enhancing gradient flow.
- **Model Architecture**: Similar to the Keras version but uses custom layers for flexibility.
- **Compilation and Training**: Identical to the Keras approach, ensuring consistency.

**Theories and Concepts**

- **Supervised Learning**: The model learns a mapping from TF-IDF features to category labels using labeled data.
- **Transformer Architecture**: Introduced in "Attention is All You Need" (Vaswani et al., 2017), Transformers use self-attention to model dependencies in sequential data. Here, we adapt it for TF-IDF vectors treated as a sequence of length 1.
- **Multi-Head Attention**: Allows the model to focus on different parts of the input simultaneously, capturing diverse relationships:
  $$
  \text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \ldots, \text{head}_h)W^O
  $$
  where each $\text{head}_i = \text{Attention}(QW_i^Q, KW_i^K, VW_i^V)$.
- **Residual Connections**: Add the input to the output of a layer to improve gradient flow and training stability.
- **Layer Normalization**: Normalizes activations to stabilize and accelerate training:
  $$
  \text{LayerNorm}(x) = \gamma \cdot \frac{x - \mu}{\sqrt{\sigma^2 + \epsilon}} + \beta
  $$
- **Early Stopping**: Monitors validation loss to stop training when performance plateaus, preventing overfitting.

**Model Evaluation Approach**

The model is evaluated on the test set using multiple metrics to ensure robust performance:
- **Accuracy**: Proportion of correct predictions.
- **Precision, Recall, F1-Score**: Macro-averaged to account for class imbalance, measuring per-class performance:
  $$
  \text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}}, \quad \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}}, \quad \text{F1} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
  $$
- **ROC-AUC**: Area under the Receiver Operating Characteristic curve, measuring class separability (one-vs-rest for multi-class).
- **Precision-Recall Curve**: Plots precision vs. recall for each class, with AUC indicating performance on imbalanced data.
- **Confusion Matrix**: Visualizes true vs. predicted labels to identify misclassifications.
- **Decision Boundary Visualization**: Uses PCA to reduce features to 2D, plotting decision regions and true/predicted labels.
- **Word Cloud**: Visualizes frequent terms in the dataset to highlight key features.

**Theories and Concepts**

- **Confusion Matrix**: A table showing true positives (TP), true negatives (TN), false positives (FP), and false negatives (FN) for each class.
- **ROC-AUC**: Measures the model's ability to distinguish classes, with higher AUC indicating better performance.
- **Precision-Recall Curve**: Focuses on positive class performance, critical for imbalanced datasets.
- **PCA (Principal Component Analysis)**: Reduces high-dimensional data to 2D for visualization while preserving variance:
  $$
  X_{\text{PCA}} = XW, \quad \text{where } W \text{ are principal components}
  $$

## Longest Sentence Length

**Analyze Sentence Length**

This step analyzes the length of the product descriptions to understand the data's scale and to inform decisions about text processing.

- The code defines a function to find the longest sentence in the Text column of a given DataFrame.
- It then applies this function to both the training and testing sets to determine the character and word counts of their respective longest sentences. This information can be useful for tasks like setting maximum sequence lengths for deep learning models.

```python
def find_longest_sentence_and_word_count(df, column_name):
    longest_sentence_text = None
    max_char_length = 0
    longest_sentence_word_count = 0

    if column_name in df.columns:
        for text_entry in df[column_name].dropna():
            if isinstance(text_entry, str):
                current_char_length = len(text_entry)
                if current_char_length > max_char_length:
                    max_char_length = current_char_length
                    longest_sentence_text = text_entry
                    longest_sentence_word_count = len(text_entry.split())
        
        if longest_sentence_text:
            return longest_sentence_text, max_char_length, longest_sentence_word_count
        else:
            return None, 0, 0
    else:
        print(f"Error: '{column_name}' column not found in the DataFrame.")
        return None, 0, 0

print("Finding longest sentence in 'Text' column of train_df...")
longest_train_sentence, train_char_count, train_word_count = find_longest_sentence_and_word_count(train, 'Text')
print(f"Longest sentence in train_df['Text'] (character length: {train_char_count}, word count: {train_word_count}):")

print("="*40 + "\n")

print("Finding longest sentence in 'Text' column of test_df...")
longest_test_sentence, test_char_count, test_word_count = find_longest_sentence_and_word_count(test, 'Text')
print(f"Longest sentence in test_df['Text'] (character length: {test_char_count}, word count: {test_word_count}):")

print("="*40 + "\n")
```

## Text Cleaning

**Text Preprocessing**

A function called preprocess_text is defined to standardize the raw text data. This function performs a series of transformations:
- All text is converted to lowercase to ensure consistency.
- Punctuation and numbers are removed, leaving only alphabetic characters.
- The cleaned text is split into individual words (tokenization).
- Common stop words (like "the", "is", "a") are removed, as they generally don't contribute to the meaning of a document.
- Finally, Porter stemming is applied to reduce each word to its root form (e.g., "running" becomes "run"), which helps in grouping similar words together.

This cleaned text is then used to replace the original Text column in both the training and testing DataFrames.

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

def preprocess_text(text):
    if not isinstance(text, str):
        return ""

    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    words = text.split()
    processed_words = [stemmer.stem(word) for word in words if word not in stop_words]
    
    return ' '.join(processed_words)
    
train['Text'] = train['Text'].apply(preprocess_text)
test['Text'] = test['Text'].apply(preprocess_text)
train
```

# Features Engineering

**Feature and Target Split**

The cleaned datasets are now split into their respective components for machine learning. The Text column, which contains the product descriptions, is separated into features (`x_train, x_test`), while the Label column, which contains the categories, is separated into target labels (`y_train, y_test`). This is a standard practice that prepares the data for model training and evaluation.

```python
x_train = train.drop('Label', axis=1) 
y_train = train['Label'] 
x_test = test.drop('Label', axis=1) 
y_test = test['Label'] 

print("X_train shape:", x_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", x_test.shape)
print("y_test shape:", y_test.shape)
```

## Text Representation (TF-IDF)

**Text Representation**

Since machine learning models require numerical input, the text features are converted into a numerical format using TF-IDF (Term Frequency-Inverse Document Frequency). The `TfidfVectorizer` creates a matrix where each row represents a document and each column represents a unique word, with values indicating the importance of that word in the document. To improve computational efficiency and performance, Truncated SVD (Singular Value Decomposition) is then used to reduce the dimensionality of this TF-IDF matrix from 5000 features to just 128.

```python
tfidf_vectorizer = TfidfVectorizer(
    tokenizer=lambda x: x.split(),
    max_features=5000,
    stop_words='english',
    lowercase=True
)

x_train = tfidf_vectorizer.fit_transform(x_train['Text'])

x_test = tfidf_vectorizer.transform(x_test['Text'])

x_train = x_train.toarray()
x_test = x_test.toarray()

print(f"x_train_tfidf shape: {x_train.shape}")
print(f"x_test_tfidf shape: {x_test.shape}")
print("Sample TF-IDF features for x_train (first 5 rows, first 5 columns):\n", x_train[:5, :5])
print("="*40 + "\n")
```

## Dimensionality Reduction (TruncatedSVD)

```python
n_components = 128

print(f"Applying TruncatedSVD with n_components={n_components}...")
svd = TruncatedSVD(n_components=n_components, random_state=42)

x_train = svd.fit_transform(x_train)
x_test = svd.transform(x_test)

print(f"x_train shape after SVD: {x_train.shape}")
print(f"x_test shape after SVD: {x_test.shape}")
print("="*40 + "\n")
```

## Target Labels Encoding

**Target Label Encoding**

The final preprocessing step is to convert the categorical target labels (e.g., "Books", "Household") into a numerical format. The `LabelEncoder` is fit on the training labels to learn the unique categories and their corresponding numerical values. It then transforms both the training and testing labels into these integer representations, which is the required format for training the classification model.

```python
le = LabelEncoder()

y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)

print("\nEncoded y_train:", y_train)
print("Encoded y_test:", y_test)

print("\nClasses learned by LabelEncoder:", le.classes_)
```

# Model Training

## Model Architecture, Model Building, and Model Training (Transformer with Keras)

**Model Architecture and Training**

This model architecture constructs a Transformer architecture using the Keras Functional API, defining a sequence of layers that process the TF-IDF features. The model's key components include `MultiHeadAttention`, which allows it to weigh the importance of different terms, as well as `Dense` and `Dropout` layers for classification. After defining the architecture, the model is compiled with an Adam optimizer and `SparseCategoricalCrossentropy` loss function. Finally, the model is trained on the training data. The process uses an `EarlyStopping` callback, which automatically halts training when the validation loss stops improving to prevent the model from overfitting.

**Model Architecture, model building, and model training (Transformer with Keras)**

The first section defines key parameters that control the size and behavior of the model.

- `feature_dim`: This is the input size, which corresponds to the number of TF-IDF features (128 after SVD).
- `num_heads`: Sets the number of attention heads in the `MultiHeadAttention` layer, allowing the model to focus on different parts of the input text simultaneously.
- `ff_dim`: Defines the size of the hidden layer within the feed-forward network, which adds complexity to the model.
- `dropout_rate`: A regularization technique that randomly drops connections during training to prevent overfitting.
- `num_classes`: The number of output categories (four in this case).

**Building the Model**

The model is constructed using the Keras Functional API, which provides a flexible way to build complex neural networks.

- `Input Layer`: The model starts with an `Input` layer that expects the TF-IDF vectors.
- `Reshaping`: The TF-IDF vector is reshaped to have an extra dimension. This allows it to be treated as a sequence of length 1, which is the required format for the Transformer's attention layer.
- `Transformer Block`: This is the core of the model. It consists of two main parts:
    - `Multi-Head Attention`: This layer allows the model to weigh the importance of different features. It can effectively learn which combinations of words are most relevant for classification.
    - `Feed-Forward Network`: This simple network adds non-linearity and further processes the output of the attention mechanism.
- `Residual Connections and Normalization`: The outputs of both the attention and feed-forward layers are combined with their original inputs via "residual connections" to help with training deep networks. `LayerNormalization` is also applied to stabilize the learning process.
- `Global Pooling`: The output of the Transformer block is condensed into a single vector using `GlobalAveragePooling1D`, which prepares the data for the final classification layers.
- `Classification Head`: Two final Dense layers are added to perform the classification. The last layer uses a softmax activation function to output a probability distribution over the four categories.

**Model Compilation and Training**

Once the architecture is defined, the model is prepared for training.

```python
feature_dim = x_train.shape[1]
num_heads = 4
ff_dim = 128
dropout_rate = 0.1
num_classes = len(le.classes_)

if feature_dim % num_heads != 0:
    print(f"Warning: feature_dim ({feature_dim}) is not divisible by num_heads ({num_heads}).")
    print("Consider adjusting num_heads or feature_dim to ensure divisibility.")

inputs = keras.Input(shape=(feature_dim,), name="tfidf_input")

x = layers.Reshape((1, feature_dim))(inputs)

attn_output = layers.MultiHeadAttention(num_heads=num_heads, key_dim=feature_dim // num_heads)(x, x, x)

attn_output = layers.Dropout(dropout_rate)(attn_output)
out1 = layers.LayerNormalization(epsilon=1e-6)(x + attn_output)

ffn_output = layers.Dense(ff_dim, activation='relu')(out1)
ffn_output = layers.Dropout(dropout_rate)(ffn_output)

transformer_output = layers.LayerNormalization(epsilon=1e-6)(out1 + ffn_output)

x = layers.GlobalAveragePooling1D()(transformer_output)

x = layers.Dense(64, activation='relu')(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(num_classes, activation='softmax', name="output_layer")(x)

model = keras.Model(inputs=inputs, outputs=outputs, name="transformer_text_classifier")

print(f"\n--- Keras Transformer Model Architecture for Text Classification ---")
model.summary()
print("="*40 + "\n")

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss=keras.losses.SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)

num_epochs = 200
patience = 5

early_stopping_callback = EarlyStopping(
    monitor='val_loss',
    patience=patience,
    restore_best_weights=True,
    verbose=1
)

print("--- Starting Training with Early Stopping ---")
history = model.fit(
    x_train, y_train,
    epochs=num_epochs,
    validation_split=0.2,
    callbacks=[early_stopping_callback],
    verbose=1
)
print("\n--- Training Complete ---")
print("="*40 + "\n")
```

## Model Architecture, Model Building, and Model Training (TensorFlow custom layers and functions)

This model builds and trains a text classification model using a custom implementation of a Transformer architecture in TensorFlow. Instead of using Keras's built-in layers for the core Transformer components, this approach defines its own classes for the attention mechanism and the full Transformer block.

**MultiHeadSelfAttention Layer**

This custom class is the heart of the Transformer. It's a self-contained layer that calculates attention scores, which allow the model to learn the relationships and relative importance of different features (in this case, the TF-IDF features). It includes:

- **Projection Layers:** Dense layers to transform the input features into three different representations: `query, key, and value`.
- **Head Separation:** Logic to split these projections into multiple "heads," which allows the model to attend to different parts of the input simultaneously.
- **Attention Calculation:** A function that performs a scaled dot-product attention calculation. It takes the queries and keys, calculates a similarity score, and uses those scores to create a weighted sum of the values.

**TransformerBlock Layer**

This class acts as the main building block for the Transformer model. It combines multiple components into a single, reusable layer.

- It incorporates the custom `MultiHeadSelfAttention` layer.
- It contains a feed-forward neural network to add more complexity to the model after the attention mechanism.
- It includes `LayerNormalization` layers and residual connections (the addition of the input to the output). These are critical for training deep networks, as they help prevent the model from getting stuck in training and allow for stable learning.

**Final Model Architecture**

The overall model is built using the custom `TransformerBlock` as its central feature extractor.

- The TF-IDF features are passed to the model's input layer. The features are then reshaped from $\text{(batch\_size, feature\_dim)}$ to $\text{(batch\_size, 1, feature\_dim)}$ and fed into the custom `TransformerBlock`. 
- `TransformerBlock`: This block performs the attention calculation across the features and feature transformation. The output of the block is passed through a Lambda layer that squeezes (removes) the sequence dimension, resulting in a vector of shape $\text{(batch\_size, feature\_dim)}$, which is then passed to a Dense layer for classification, with a final softmax activation that outputs the predicted probabilities for each of the four categories.

**Compilation and Training**

Finally, the custom model is compiled and trained.

- **Model Complier:** It's configured with the Adam optimizer and the SparseCategoricalCrossentropy loss function.
- **Training:** The model is trained on the data with an `EarlyStopping` callback, which automatically monitors the validation loss and stops training once the model's performance plateaus. This is a crucial step to prevent overfitting and ensure the final model generalizes well to new data.

```python
class MultiHeadSelfAttention(layers.Layer):
    def __init__(self, embed_dim, num_heads=8, **kwargs):
        super(MultiHeadSelfAttention, self).__init__(**kwargs)
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        
        if embed_dim % num_heads != 0:
            raise ValueError(
                f"embedding dimension = {embed_dim} should be divisible by number of heads = {num_heads}"
            )
        self.projection_dim = embed_dim // num_heads

        self.query_dense = layers.Dense(embed_dim)
        self.key_dense = layers.Dense(embed_dim)
        self.value_dense = layers.Dense(embed_dim)
        
        self.combine_heads = layers.Dense(embed_dim)

    def attention(self, query, key, value):
        score = tf.matmul(query, key, transpose_b=True)
        dim_key = tf.cast(tf.shape(key)[-1], tf.float32)
        scaled_score = score / tf.math.sqrt(dim_key)
        weights = tf.nn.softmax(scaled_score, axis=-1)
        output = tf.matmul(weights, value)
        return output, weights

    def separate_heads(self, x, batch_size):
        x = tf.reshape(x, (batch_size, -1, self.num_heads, self.projection_dim))
        return tf.transpose(x, perm=[0, 2, 1, 3])

    def call(self, inputs):
        batch_size = tf.shape(inputs)[0]

        query = self.query_dense(inputs)
        key = self.key_dense(inputs)
        value = self.value_dense(inputs)

        query = self.separate_heads(query, batch_size)
        key = self.separate_heads(key, batch_size)
        value = self.separate_heads(value, batch_size)

        attention, weights = self.attention(query, key, value)
        
        attention = tf.transpose(attention, perm=[0, 2, 1, 3])
        concat_attention = tf.reshape(attention, (batch_size, -1, self.embed_dim))
        
        output = self.combine_heads(concat_attention)
        return output

    def get_config(self):
        config = super(MultiHeadSelfAttention, self).get_config()
        config.update({
            "embed_dim": self.embed_dim,
            "num_heads": self.num_heads,
        })
        return config

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, rate=0.1, **kwargs):
        super(TransformerBlock, self).__init__(**kwargs)
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.ff_dim = ff_dim
        self.rate = rate

        self.att = MultiHeadSelfAttention(embed_dim=embed_dim, num_heads=num_heads)
        
        self.ffn = keras.Sequential(
            [layers.Dense(ff_dim, activation="relu"), layers.Dense(embed_dim),]
        )
        
        self.layernorm1 = layers.LayerNormalization(epsilon=1e-6)
        self.layernorm2 = layers.LayerNormalization(epsilon=1e-6)
        
        self.dropout1 = layers.Dropout(rate)
        self.dropout2 = layers.Dropout(rate)

    def call(self, inputs, training=None):
        attn_output = self.att(inputs)
        attn_output = self.dropout1(attn_output, training=training)
        out1 = self.layernorm1(inputs + attn_output)

        ffn_output = self.ffn(out1)
        ffn_output = self.dropout2(ffn_output, training=training)
        return self.layernorm2(out1 + ffn_output)

    def get_config(self):
        config = super(TransformerBlock, self).get_config()
        config.update({
            "embed_dim": self.embed_dim,
            "num_heads": self.num_heads,
            "ff_dim": self.ff_dim,
            "rate": self.rate,
        })
        return config

feature_dim = n_components
num_heads = 4
ff_dim = 128
dropout_rate = 0.1

if feature_dim % num_heads != 0:
    print(f"Warning: feature_dim ({feature_dim}) is not divisible by num_heads ({num_heads}).")
    print("Consider adjusting num_heads or feature_dim to ensure divisibility.")

inputs = keras.Input(shape=(feature_dim,), name="tfidf_input")

x = layers.Reshape((1, feature_dim))(inputs)

transformer_output = TransformerBlock(
    embed_dim=feature_dim, num_heads=num_heads, ff_dim=ff_dim, rate=dropout_rate
)(x)

x = layers.Lambda(lambda t: tf.squeeze(t, axis=1))(transformer_output)

x = layers.Dense(64, activation='relu')(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(num_classes, activation='softmax', name="output_layer")(x)

model = keras.Model(inputs=inputs, outputs=outputs, name="transformer_text_classifier")

print(f"\n--- Keras Transformer Model Architecture for Text Classification ---")
model.summary()
print("="*40 + "\n")

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss=keras.losses.SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)

num_epochs = 200
patience = 5

early_stopping_callback = EarlyStopping(
    monitor='val_loss',
    patience=patience,
    restore_best_weights=True,
    verbose=1
)

print("--- Starting Training with Early Stopping ---")
history = model.fit(
    x_train, y_train,
    epochs=num_epochs,
    validation_split=0.2,
    callbacks=[early_stopping_callback],
    verbose=1
)
print("\n--- Training Complete ---")
print("="*40 + "\n")
```

## Precision-Recall Curve

**Plotting the Precision-Recall Curve**

The code plots a Precision-Recall Curve (PR Curve)  for each class. This curve illustrates the trade-off between precision and recall at different classification thresholds. The AUC (Area Under the Curve) value for each curve provides a single score to summarize performance for that class.

```python
print("\n--- Plotting Precision-Recall Curve ---")

class_labels_pr = [str(i) for i in range(num_classes)]

plt.figure(figsize=(20, 10))

sns.set_palette("Set2")

for i, label_str in enumerate(class_labels_pr):
    class_val = int(label_str)

    y_test_bin = np.where(y_true_test_np == class_val, 1, 0)

    y_pred_score = y_pred_proba_test_np[:, class_val]

    precision, recall, _ = precision_recall_curve(y_test_bin, y_pred_score)
    pr_auc = auc(recall, precision)

    plt.plot(recall, precision, label=f'Class {label_str} (AUC = {pr_auc:.2f})')

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve for Each Class (One-vs-Rest)")
plt.legend(loc="lower right")
plt.ylim([0, 1.05])
plt.xlim([0, 1])
plt.grid(True)

plt.show()
```

**Result Interpretation**

The results show that the model is performing very well. With a high overall accuracy of 94.43 percent, the model is correctly classifying most of the test data. The macro averages for precision ($\text{0.946}$), recall ($\text{0.944}$), and F1-score ($\text{0.945}$) are all high and consistent, suggesting that the model's performance is balanced across all classes. The high AUC-ROC score of $\text{0.990}$ and the high AUC for individual classes in the Precision-Recall curve (all above $\text{0.96}$) further confirm that the model is excellent at distinguishing between the different categories.

## Decision Boundary Visualization (Latent Space)

**Feature Extraction and Dimensionality Reduction**

This code visualizes the model's decision-making process by mapping its complex, high-dimensional features into a simple 2D space. A temporary sub-model is first created to extract the output from the `lambda` layer, which represents the model's learned features just before the final classification head. Principal Component Analysis (PCA) is then applied to these features to reduce their dimensionality to just two, making them plottable. This process is performed on the unseen test data to ensure the visualization accurately reflects the model's performance on new data.

**Plotting the Decision Boundary**

The code then creates a visualization to show how the model divides the 2D space to classify data points. It generates a grid of points, transforms them into the 2D PCA space, and then passes them back to the model's classification head to get predictions. These predictions are used to create a colored contour plot, which represents the decision boundary. The actual test data points are overlaid on this plot, colored by their predicted and true classes.

```python
print("\n--- Visualizing Transformer Latent Space (PCA of Lambda layer output) on Test Data ---")

PCA_FEATURE_LAYER_NAME = "lambda"
DENSE_64_LAYER_NAME = "dense_8"
OUTPUT_LAYER_NAME = "output_layer"

feature_extractor_for_pca = keras.Model(
    inputs=model.inputs,
    outputs=model.get_layer(PCA_FEATURE_LAYER_NAME).output
)

input_to_dense_64 = model.get_layer(DENSE_64_LAYER_NAME).input
output_layer_for_head = model.get_layer(OUTPUT_LAYER_NAME).output

classification_head_model = keras.Model(
    inputs=input_to_dense_64,
    outputs=output_layer_for_head
)

intermediate_features_test = feature_extractor_for_pca.predict(x_test)
true_labels_test = y_test

pca = PCA(n_components=2)
pca.fit(intermediate_features_test)
pca_features_test = pca.transform(intermediate_features_test)

predicted_labels_test = np.argmax(model.predict(x_test), axis=1)

x_min, x_max = pca_features_test[:, 0].min() - 1, pca_features_test[:, 0].max() + 1
y_min, y_max = pca_features_test[:, 1].min() - 1, pca_features_test[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))

meshgrid_points_2d = np.c_[xx.ravel(), yy.ravel()]

reconstructed_high_dim_features = pca.inverse_transform(meshgrid_points_2d)

outputs_meshgrid = classification_head_model.predict(reconstructed_high_dim_features)
Z = np.argmax(outputs_meshgrid, axis=1)
Z = Z.reshape(xx.shape)

cmap_regions = plt.cm.get_cmap('viridis', num_classes)
cmap_scatter = sns.color_palette("tab10", n_colors=num_classes)

fig_db, axes_db = plt.subplots(1, 2, figsize=(24, 10))

ax_db1 = axes_db[0]
contourf_plot = ax_db1.contourf(xx, yy, Z, alpha=0.4, cmap=cmap_regions)
sns.scatterplot(
    x=pca_features_test[:, 0],
    y=pca_features_test[:, 1],
    hue=predicted_labels_test,
    palette=cmap_scatter,
    legend='full',
    alpha=0.7,
    s=50,
    ax=ax_db1
)
ax_db1.set_title('Transformer Decision Boundary (Predicted Classes on Test Data)', fontsize=16)
ax_db1.set_xlabel('Principal Component 1', fontsize=12)
ax_db1.set_ylabel('Principal Component 2', fontsize=12)
ax_db1.grid(True, linestyle='--', alpha=0.6)
ax_db1.set_xlim(xx.min(), xx.max())
ax_db1.set_ylim(yy.min(), yy.max())

ax_db2 = axes_db[1]
sns.scatterplot(
    x=pca_features_test[:, 0],
    y=pca_features_test[:, 1],
    hue=true_labels_test,
    palette=cmap_scatter,
    legend='full',
    alpha=0.7,
    s=50,
    ax=ax_db2
)
ax_db2.set_title('Transformer Latent Space (True Classes on Test Data)', fontsize=16)
ax_db2.set_xlabel('Principal Component 1', fontsize=12)
ax_db2.set_ylabel('Principal Component 2', fontsize=12)
ax_db2.grid(True, linestyle='--', alpha=0.6)
ax_db2.set_xlim(xx.min(), xx.max())
ax_db2.set_ylim(yy.min(), yy.max())

fig_db.suptitle('Decision Boundary in PCA Latent Space (Test Data)', fontsize=18, y=1.02)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

**Result Interpretation**

The plot provides a clear view of how well the model has learned to separate the classes in the latent space.

- The two subplots show the same data points. The first is colored by the model's predictions, while the second is colored by the true labels. A high degree of similarity between these two plots indicates that the model's predictions align very closely with the actual class distributions.
- The clear separation between the different color regions (the decision boundaries) and the fact that most points fall within their correctly colored area confirms that the model is performing well. The visualization confirms that the model's high accuracy and F1-scores are not a fluke; it has successfully learned to create distinct, well-defined boundaries that generalize effectively to unseen data.
