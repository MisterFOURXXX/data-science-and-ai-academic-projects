## Problem-Solving

To solve this problem, I will follow a structured methodology that involves data collection, preprocessing, model development, and deployment. The first step involves gathering a dataset of brain X-ray images and their corresponding labels from the hospital's existing database of previous patients who have consented to the use of their data. The data will be validated by hospital doctors to ensure the accuracy of the labels.

The primary technique for this classification task is Convolutional Neural Networks (CNNs). CNNs are a class of deep neural networks specifically designed for processing structured grid-like data, such as images. Their architecture is particularly well-suited for image classification due to their ability to automatically learn spatial hierarchies of features from the input data.

- **Convolutional Layer:** This is the foundational building block of a CNN. It applies a set of learnable filters (or kernels) to the input image, creating feature maps that highlight specific features like edges, textures, and shapes. The theory behind this is that by sliding a small filter over the input, the network can detect the same features in different locations, making it translation invariant. The activation function (e.g., ReLU) introduces non-linearity, allowing the network to learn more complex patterns.
- **Pooling Layer:** Following the convolutional layer, a pooling layer (e.g., max pooling) is used to reduce the spatial dimensions of the feature maps. This reduces the number of parameters and computation, and helps make the detected features more robust to small shifts or distortions in the input image. This process is based on the idea of downsampling, which summarizes the features present in a local region of the feature map.
- **Fully Connected Layer:** After several convolutional and pooling layers, the high-level features learned by the network are flattened into a single vector and fed into a standard neural network (a fully connected layer). This part of the network uses these features to perform the final classification.

CNN model will be trained on the prepared dataset to classify each X-ray film into the correct tumor stage. After training, the model will be deployed on the hospital's server, enabling nurses and doctor's assistants to use it for initial patient screening. The results from the model will then guide doctors in providing more focused and efficient patient care.

# About experiments

**Experimental Design**

The core of the experimental phase is to fine-tune the CNN model to achieve the best possible performance. This involves a series of experiments focusing on various aspects of the machine learning pipeline and model architecture. The theories behind these adjustments are crucial for optimizing performance:

**Oversampling/Undersampling:**

These techniques address class imbalance, a common issue in medical datasets where one class (e.g., 'No Tumor') may have significantly more examples than others.

  - Oversampling (e.g., SMOTE) creates synthetic examples of the minority class, helping the model learn from a balanced distribution.
  - Undersampling removes examples from the majority class. The underlying theory is to prevent the model from becoming biased towards the dominant class, thereby improving its ability to correctly classify all classes.

**Model Architecture:** 

The design of the CNN architecture is critical. Experiments will involve adjusting:
  - **Number of Filters:** Increasing the number of filters allows the model to learn more complex and diverse features.
  - **Kernel Size:** The size of the filter affects the receptive field of the convolution, with larger kernels capturing broader features.
  - **Pooling Size:** Adjusting pooling size balances between feature summarization and information loss.
  - **Number of Neurons:** The number of neurons in the fully connected layers determines the model's capacity to learn from the features extracted by the convolutional layers.

**Hyperparameter Tuning:**

- **Learning Rate Scheduler:** This technique dynamically adjusts the learning rate during training. A high initial learning rate can help the model converge faster, while a lower rate at later stages can help it fine-tune the weights for better accuracy.
- **Batch Size:** The number of training examples used in each iteration. A smaller batch size can introduce more noise but can help the model escape local minima, while a larger batch size provides a more stable estimate of the gradient.

**Activation Functions:**

- **Convolutional Layers:** Functions like `ReLU` (`f(x)=max(0,x)`) introduce non-linearity, enabling the model to learn complex relationships in the data.
- **Dense Layers:** For multi-class classification, a Softmax activation function is used in the final dense layer. Softmax converts the raw output scores into a probability distribution, with each value representing the probability that the input belongs to a specific class. This is based on the principle of normalized exponential function, which ensures that the sum of all output probabilities is 1.

The best-performing model will be selected based on a comprehensive evaluation using metrics such as Accuracy, Confusion Matrix, Precision, Recall, F1-Score, and Area Under the ROC Curve (AUC-ROC). These metrics provide a holistic view of the model's performance, considering both its overall correctness and its ability to handle different types of classification errors.

# Load images

The first step in the data pipeline is to load the image files and their corresponding labels. A custom function is defined to traverse the directory structure, reading each image and extracting its folder name as the label. This process effectively converts the file system structure into a structured dataset.

The labels are then renamed for clarity using a predefined mapping. For instance, 'glioma' is changed to 'Glioma' and so on. This simple step is an important practice in data preparation to ensure consistency and readability.

A sample of the combined dataset is visualized to confirm that the images have been loaded correctly and are associated with their proper labels. This visual check is a form of sanity check, a fundamental practice in data science to verify that the initial data ingestion has worked as expected.

```python
def load_images_from_folder(image_dir):
    images = []
    labels = []
    for root, _, files in os.walk(image_dir):
        for file in files:
            if file.endswith(('.jpg')):
                image_path = os.path.join(root, file)
                img = cv2.imread(image_path)
                images.append(img)
                labels.append(os.path.basename(root))
    return images, labels

# Define dataset directory
image_dir = '/kaggle/input/brain-tumor-mri-dataset'

# Load training and testing data
train_images, train_labels = load_images_from_folder(os.path.join(image_dir, 'Training'))
test_images, test_labels = load_images_from_folder(os.path.join(image_dir, 'Testing'))
images = train_images + test_images
labels = train_labels + test_labels
```

```python
class_mapping = {'glioma': 'Glioma', 'meningioma': 'Meninginoma', 'notumor': 'No Tumor', 'pituitary': 'Pituitary'}
new_labels = [class_mapping[label] for label in labels]
labels = np.array(new_labels)
```

## Resize images

The first step in the pipeline is to resize all images to a uniform dimension. This is a crucial preprocessing step for two main reasons:

**Standardized Input:**

All images must have the same size to form a coherent input tensor for a neural network. A Convolutional Neural Network (CNN) processes input data as a fixed-size grid (tensor), and variations in image size would break this structure.

**Computational Efficiency:** 

Larger images require more memory and computational resources, increasing training time and potentially causing memory errors. By resizing images to a smaller, more manageable size, we can significantly reduce the computational load while retaining essential visual information.

Based on the Exploratory Data Analysis (EDA), images in the dataset have varying dimensions, with an average size of 512x512 pixels. To balance model performance and computational constraints, a target size of 256x256 pixels was chosen. While a larger size might capture more detail, a smaller size is more computationally feasible and less prone to issues like notebook crashes. The resizing operation is performed using `cv2.resize()`, which applies an interpolation algorithm to resample the image pixels.

## Normalized images

Normalization is a standard practice in deep learning that scales input features to a small, consistent range, typically between 0 and 1. This is a key step for several theoretical reasons:

- Improved Training Stability: Neural networks are trained using gradient-based optimization algorithms. Without normalization, large pixel values could lead to large gradients, causing unstable training, slow convergence, and potentially vanishing or exploding gradients.
- Faster Convergence: Normalizing the input data ensures that all features contribute equally to the loss function. This helps the optimization algorithm find the optimal solution more quickly.

The images, with pixel values ranging from 0 to 255, are normalized by simply dividing each pixel by 255. This linear transformation scales the data to the desired [0, 1] range, as confirmed by checking the minimum and maximum pixel values after the operation.

The class labels ('Glioma', 'Meningioma', etc.) are categorical and cannot be used directly by a neural network. They must be converted into a numerical format. This is a two-step process:

- Label Encoding: The string labels are first converted into integer labels (e.g., 'Glioma' becomes 0, 'Meningioma' becomes 1, etc.). This is done using `LabelEncoder` from scikit-learn.
- One-Hot Encoding: The integer labels are then transformed into a binary matrix using `to_categorical()` from Keras. This is necessary because the model's output will be a probability distribution over the classes. For example, the label for 'Glioma' (encoded as 0) will be represented as `[1, 0, 0, 0]`.

This process is vital because the chosen loss function, categorical cross-entropy, requires the labels to be in a one-hot encoded format. The theory behind this is that categorical cross-entropy measures the performance of a classification model whose output is a probability value between 0 and 1. The one-hot encoded labels provide the "correct" probability distribution for the loss function to compare against the model's predictions.

## Initial convolutional neural network

The model is designed as a sequential stack of layers, a standard approach for building simple neural networks. The architecture follows a pattern of alternating convolutional and pooling layers to progressively extract and summarize features from the input images, followed by fully connected layers for classification.

### Theoretical Foundations

**Conv2D Layer:**

This layer applies a set of learnable filters (also known as kernels) to the input image. Each filter slides over the image, performing a dot product with the input data it's currently on, and generates a feature map. The use of multiple filters allows the network to learn a variety of features, such as edges, textures, and shapes. The `ReLU` (Rectified Linear Unit) activation function (`f(x)=
max(0,x)`) is applied after each convolution. This non-linear function is computationally efficient and helps the model learn complex, non-linear relationships in the data.

**MaxPooling2D Layer:** 

This layer is a form of down-sampling that reduces the spatial dimensions of the feature maps. It works by taking the maximum value from a specified window. The theoretical benefit of max pooling is that it reduces the number of parameters and computational cost while making the network more robust to minor shifts or distortions in the input image.

**Flatten Layer:**

After the convolutional and pooling layers, the high-dimensional feature maps are converted into a one-dimensional vector. This step is necessary to bridge the output of the convolutional layers with the input of the fully connected layers.

**Dense Layer:** 

These are standard neural network layers where every neuron is connected to every neuron in the previous layer. The first dense layer acts as a classifier, interpreting the high-level features extracted by the convolutional layers. The final dense layer, with four neurons (one for each class), uses the Softmax activation function. Softmax converts the raw output scores into a probability distribution, where the sum of all probabilities is 1.0. This allows the model to output a confidence score for each class, making it easy to determine the most likely tumor type.

### Model architecture

```python
# Define the CNN model
def create_cnn_model(input_shape, num_classes):
    model = Sequential()

    # Convolutional layer 1
    model.add(Conv2D(32, (3, 3), activation='relu', input_shape=input_shape))
    model.add(MaxPooling2D(pool_size=(2, 2)))

    # Convolutional layer 2
    model.add(Conv2D(64, (3, 3), activation='relu')) 
    model.add(MaxPooling2D(pool_size=(2, 2)))

    # Convolutional layer 3
    model.add(Conv2D(128, (3, 3), activation='relu'))
    model.add(MaxPooling2D(pool_size=(2, 2)))

    # Flatten layer
    model.add(Flatten())

    # Fully connected layer
    model.add(Dense(128, activation='relu'))

    # Output layer
    model.add(Dense(num_classes, activation='softmax'))  # Use softmax for multi-class classification

    return model

# Dataset shapes
input_shape = (256, 256, 3)  # Shape of each image
num_classes = 4              # Number of classes in the output

# Create the CNN model
cnn_model = create_cnn_model(input_shape, num_classes)

cnn_model.summary()
```

**Loss Function:**

### Early Stopping

To prevent overfitting, a common problem in deep learning where a model performs well on training data but poorly on unseen data, Early Stopping is implemented.

**Theoretical**

Early stopping is a form of regularization. It monitors the model's performance on a validation set during training. If the validation loss does not improve for a specified number of epochs (in this case, 5), the training process is halted, and the model's weights are restored to the point where the validation loss was at its minimum. This prevents the model from continuing to learn from the training data to the point where it memorizes noise rather than learning generalizable patterns.

```python
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=5, 
    restore_best_weights=True )
```

Early stopping
- Monitor loss in validation dataset.
- If the value of loss in validation dataset did not improve in 5 epoch, The model will stop training.

## Final convolutional neural network

Based on the extensive hyperparameter tuning and experimentation, a final Convolutional Neural Network (CNN) architecture was constructed. This model incorporates the optimal configurations identified in the preceding experiments, along with additional regularization techniques to enhance its performance and prevent overfitting. The final model is a culmination of the insights gained, balancing computational efficiency with predictive power.

### Model Architecture and Components

The final model is a sequential deep learning architecture, meaning the layers are stacked in a linear fashion. It follows a classic CNN design, consisting of three convolutional blocks followed by a fully connected classifier.

- **Convolutional Blocks:** Each block is composed of a `Conv2D` layer and a `MaxPooling2D` layer.
  - **Conv2D Layers:** The number of filters increases with each layer `(32, 64, 128)` following the principle that deeper layers learn more complex, abstract features. The kernel size for all layers is set to `(6,6)`, as this was found to be the most effective at capturing salient features in the images.
  - **MaxPooling2D Layers:** The pooling size is set to `(4,4)`, which aggressively downsamples the feature maps. This reduces the number of parameters and makes the model more robust to minor shifts and distortions in the input images. All convolutional layers use the `ReLU` activation function, which prevents vanishing gradients and is computationally efficient.

- **Flatten Layer:** This layer transforms the 3D feature maps from the last convolutional block into a 1D vector, preparing the data for the fully connected layers

- **Fully Connected Layers:**
  - **Dense Layer:** A single hidden dense layer with 128 neurons and a ReLU activation function was chosen. This configuration provides the optimal model capacity without leading to overfitting.
  - **Batch Normalization:** This layer is added after the dense layer. It normalizes the inputs to the next layer, preventing the exploding or vanishing gradient problems. It stabilizes the learning process and allows for higher learning rates, speeding up convergence.
  - **Dropout:** A dropout layer with a rate of 0.3 is included. Dropout is a powerful regularization technique where a random percentage (30 percent) of neurons in a layer are temporarily ignored during training. This forces the network to learn more robust features and prevents it from becoming too reliant on specific neurons, thereby mitigating overfitting.

- **Output Layer:** The final layer is a dense layer with four neurons, corresponding to the four target classes. The `Softmax` activation function is used to produce a probability distribution over the classes, where the sum of probabilities for each image is 1.

```python
def create_cnn_model(input_shape, num_classes):
    model = Sequential()

    # Convolutional layer 1
    model.add(Conv2D(32, (6, 6), activation='relu', input_shape=input_shape))
    model.add(MaxPooling2D(pool_size=(4, 4)))

    # Convolutional layer 2
    model.add(Conv2D(64, (6, 6), activation='relu'))
    model.add(MaxPooling2D(pool_size=(4, 4)))

    # Convolutional layer 3
    model.add(Conv2D(128, (6, 6), activation='relu'))
    model.add(MaxPooling2D(pool_size=(4, 4)))

    # Flatten layer
    model.add(Flatten()) 
    
    # Fully connected layer
    model.add(Dense(128, activation='relu'))
    model.add(tf.keras.layers.BatchNormalization())
    model.add(Dropout(0.3)) 

    # Output layer
    model.add(Dense(num_classes, activation='softmax'))  

    return model

# Dataset shapes
input_shape = (256, 256, 3)  # Shape of each image
num_classes = 4  # Number of classes


# Create the CNN model
cnn_model = create_cnn_model(input_shape, num_classes)

# Print model summary
cnn_model.summary()
```

### Model Compilation and Training

The final model was compiled with the best-performing hyperparameters from the previous experiments.

- **Optimizer:** The Adam optimizer was selected. It was configured with a learning rate scheduler to decay the learning rate over time, which helps the model fine-tune its weights and converge to a better solution. Additionally, gradient clipping (specifically `clipnorm=1.0`) was implemented to prevent exploding gradients, where the gradient values become excessively large, causing training to diverge. This acts as an additional form of regularization.
- **Loss Function:** Categorical cross-entropy was used as the loss function, which is suitable for multi-class classification tasks with one-hot encoded labels.
- **Metrics:** Accuracy was used to monitor the model's performance during training.

Training was conducted with a batch size of 128 and an early stopping callback. The early stopping was set to monitor the validation loss with a patience of 8 epochs, meaning training would halt if the validation loss did not improve for eight consecutive epochs. This prevents the model from overfitting and saves computational resources.

```python
# Schedule e learning rate
lr_schedule = optimizers.schedules.ExponentialDecay(
        initial_learning_rate=1e-4,
        decay_steps=1500,
        decay_rate=0.9,
        staircase=True)
```

**Compile the model**

```python
# Compile the model
cnn_model.compile(
    optimizer=optimizers.Adam(learning_rate=lr_schedule, clipnorm=1.0),
    loss='categorical_crossentropy',
    metrics=['accuracy'])
```

- The optimizer is setted `clipnorm = 1` (the gradient norm will be clip to maximum value of 1) into Adam optimizer to clip the gradient to prevent the gradient becoming to large (exploring gradient problem), which it can lead to divergence in training and improve convergence as well. It can act like regularization and prevent overfitting of the model.

**Early stopping**

```python
# Early stopping
early_stopping = EarlyStopping(
    monitor='val_loss',  
    patience=8,                 
    restore_best_weights=True)
```

- Early stoppping is setted equal to 8 and monitor on validation loss, it means that I am going to monitor validation loss for 8 epoch. If the validation loss is not improving, then the model stops the training.

**Model Training**

```python
start_time = time.time()       # Measure time for training the model

# Train the model
final_model= cnn_model.fit(
    x_train, y_train,  
    validation_split=0.2,
    epochs=200,  
    batch_size=128,  
    callbacks=[early_stopping],  
    verbose=1,
    shuffle=True)


end_time = time.time()
training_time = end_time - start_time
print("Training time:", training_time, "seconds")
```

## Evaluation Metrics

For classification tasks, a single metric like accuracy can sometimes be misleading, especially with imbalanced datasets. Therefore, a suite of metrics is used to provide a complete picture of model performance.

- **Accuracy:** The most intuitive metric, it measures the proportion of correctly classified instances out of the total.

- **Precision and Recall:** These two metrics provide a trade-off analysis. Precision (also known as Positive Predictive Value) answers the question: "Of all instances predicted as positive, how many were actually positive?" It is a measure of a model's exactness. Recall (also known as Sensitivity or True Positive Rate) answers: "Of all actual positive instances, how many did the model correctly identify?" It measures a model's completeness.

- **F1-Score:** The F1-score is the harmonic mean of precision and recall. It provides a single score that balances both metrics, making it a reliable indicator of a model's overall performance, especially when there's an uneven class distribution. The formula is:

- **Confusion Matrix:** A table that visualizes the performance of a classification model. It shows the number of true positives (TP), true negatives (TN), false positives (FP), and false negatives (FN) for each class. It's an invaluable tool for identifying where the model is making errors.

- **ROC Curve and AUC:** The Receiver Operating Characteristic (ROC) curve plots the True Positive Rate (TPR) against the False Positive Rate (FPR) at various threshold settings. A high AUC value indicates that the model is effective at distinguishing between classes.
